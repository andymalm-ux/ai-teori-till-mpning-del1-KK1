### 2.
Förklara hur OvR- och OvO-algoritmerna fungerar

- OvR (one vs rest) innebär att man tränar en binär klassificering för varje klass. Exempel kan vara att du har färger röd, grön, blå. Modellen jämför röd mot alla andra, grön mot alla andra osv.
- OvO (one vs one) innebär att man tränar en modell för varje par av klasser. Exempel röd mot grön, röd mot blå, blå mot grön.


### 3.
Förklara följande utvärderingsmått:

a) Confusion matrix
- Visar hur modellens prediktioner födelar sig jämfört med de faktiska resultaten. Vi använder oss av begreppen TP - True positiv, FP - False positiv, FN - False negative och TN - True negativ.

b) Accuracy
- Anger hur stor andel av alla prediktioner som modellen klassificerat rätt, tex av 100 prediktioner så blev 90 rätt, alltså 90%.

c) Precision
- Anger hur många av det modellen predikterade som positivt, faktiskt var positiva. Hög precision innebär få falska positiva (FP).

d) Recall
- Anger hur många av alla faktiskt positiva som modellen hittade. Hög recall innebär få falska negativa (FN).

e) 𝐹1-score
- Kombinerar precision och recall och räknar ut ett mellanvärde.

f) ROC-kurvan
- Används framför allt för binär klassificering. Visar hur modellen prestanda förändras när man ändra tröskelvärdet för det som beräknas positivt.


### 4.
Vad är precision-recall tradeoff för något?
- Det är kompromissen man gör mellan att undvika falska positiva och att hitta så många positiva fall som möjligt.


### 5.
Förklara översiktligt hur respektive modell fungerar.

a) Logistisk regression
- Modellen beräknar sannolikheten att en observation tillhör en viss klass. Den använder sedan ett tröskelvärde för att bestämma klassen. Tillexempel är ett mejl spam eller inte.

b) Support vector machines
- Modellen försöker hitta en gräns som separerar olika klasser. Modellen väljer den gräns som ger så stor marginal som möjligt till de närmasta datapunkterna.

c) Beslutsträd
- Modellen delar upp data genom en serie vilkor. Varje uppdelning leder till nya grenar tills modellen kommer fram till ett beslut.

d) Ensemble learning
- En metod för att kombinera flera modeller för att få bättre prediktioner.

e) Random forest
- En modell som använder flera beslutsträd som tränas på olika slumpmässiga urval av data och features.

f) Extra trees
- Samma som Random forest, fast är mer slumpmässig när träden skapas.


### 6.
Vad innebär det att vi kan kolla på feature importance med hjälp av trädmodeller såsom beslutsträd eller random forest?

- Det innebär att vi kan kolla vilka features i datan som modellen tycker är mest användbara för att göra sina prediktioner.

### 7.
Stina säger till Kalle på lunchsamtalet “jag vill ha högsta möjliga precision
för vår klassificieringsmodell”. Kalle funderar ett tag och säger “men vad
händer då med recall”? Vad hade du svarat? I vilka fall kan man tänka sig
vilja ha en så hög precision som möjligt? I vilka fall kan det vara dåligt?
Om vi tänker oss rättsväsendet där en slutgiltig dom kan leda till fängelse,
vad kan vi då säga om precision-recall tradeoff?

- Jag hade svarat att om vi försöker få så hög precision som möjligt, så kommer vi missa fler verkligt positiva fall. Detta för att precision och recall hänger ihop. Modellen kommer enbart säga positivt när den är väldigt säker.
Hög precision är viktigt att ha när konsekvenserna för falska positiva är kostsam. Exempel kan vara ett system som automatiskt kan blockera kunders konto.
Hög precision kan vara dåligt i fall där exempelvis ett system på en bank ska försöka upptäcka bedrägeri. Det kan då leda till att bredrägerier inte flaggas för att modellen inte är tilläckligt säker.
När det gäller rättsväsendet så är det i det fallet viktigt med hög precision pga att vi vill inte råka fängsla någon som är oskyldig.

### 9.
På sidan 209 står det “på träningsdatan använder vi .fit_transform(),
på valideringsdatan och testdatan använder vi endast .transform().” Förklara logiken bakom detta.

- fit_transform() gör två saker: den lär sig parametrarna (t.ex. medelvärde och standardavvikelse) från datan, och använder sedan dem för att skala datan. transform() gör bara den andra delen, använder parametrar som redan är inlärda. Man vill bara lära sig dessa parametrar från träningsdatan, och sedan återanvända exakt samma parametrar på validerings- och testdatan. Genom att bara använda .transform() på validering/test säkerställer man att dessa dataset behandlas precis som ny, okänd data skulle behandlas i produktion, skalad utifrån vad modellen redan lärt sig, utan att kika på svaren i förväg.
