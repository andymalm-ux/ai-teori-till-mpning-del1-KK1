### 2.
Vad menas med att en modell produktionssätts?

- Det menas att modellen görs tillgänglig att användas i en applikation på riktig data.

### 3.
Vad är scikit-learn för något? Biblioteket följer några centrala designprinciper. Vilka är dessa? Vad är estimators, predictors och transformers?

- Det är ett python-bibliotek för machine-learning. Det innehåller olika algoritmer och verktyg för hela ML-flödet.
Några designpriciper är att olika modeller används på ungefär samma sätt, exempelvis fit() och predict(). Man kan kombinera olika steg med pipeline, och samma verktyg kan användas på olika dataset och modeller.

- Estimater - Ett objekt som lär sig något från datan med metoden fit().
- LinearRegression är en estimator
- Predictors - En estimator som kan använda det den lärt sig för att göra förutsägelser med metoden predict().
Transform - Ett sätt att förändra eller förbereda data med metoderna fit() och transform().

### 4.
Vad är TensorFlow och Keras?

- Verktyg som används framförallt inom deep learning. Tensorflow är ett open-source-ramverk för framför allt neural network. Keras är ett API för att underlätta att bygga neural network.

### 5.
Kalle och Stina diskuterar maskininlärning över en lunch. Kalle säger “om
jag tränat en modell och den inte presterar bra nog på testdatan så justerar
jag den tills den gör det.” Stina säger “det är ett stort fel att göra så, det
enda du då åstadkommer är att du överanpassar testdatan. Hela syftet
med testdatan försvinner då”. Vad säger du om deras dialog?

- Stina har rätt, eftersom att det Kalle gör är att använda testdatan som en del i utvecklingen av modellen. På detta sätt får vi inte testat modellen på ny data. Indirekt så optimerar vi modellen mot en specifik data.


### 6.
Många AI/ML projekt uppnår inte de ursprungligen satta målen eller att
ens passera någon form av prototyp-stadie. Vad tror du detta beror på och
hur ska vi förhålla oss till det?

- Jag tror att det beror på att man delvis inte har tillgång till bra data, samt att man inte har en tydlig bild av vad det är man faktiskt slutligen vill åstadkomma med AI/ML. AI/ML är ett verktyg, inte en trollstav.
- Jag tror att man måste tänka på vad det finns för nytta med att bygga en modell. Kommer modellen bidra med en förbättring om man för den att fungera bra.
