### 3.
Förklara följande:

a) Vad är syftet med att dela upp data i träningsdata, valideringsdata och
testdata?

- Det är för att kunna träna, förbättra och slutligen utvärder hur bra en modell fungerar på ny data. Modellen tränas på en del av datasetet, och valideringsdatan används som ny data.
Efter man valt modell och hyperparamterar så testat modellen mot testdatan för att utvärdera hur bra modellen fungerar på ny data. 

b) Vad är k-delad korsvalidering för något?

- Det är en metod som används för att estimera prediktionsfel.

c) Vad är root mean squared error (RMSE) för något?

- Det är ett mått som används för att visa hur stora fel en modell gör i sina prediktioner.

d) Vad är en hyperparameter för något? Vad är en parameter för något?

- En hyperparameter är "hur ska vi lära oss", och parameter är "vad har vi lärt oss". Hyperparametrar är något vi som utvecklar väljer innan vi tränar och validerar en modell, medan parametrar är variabler som lärts under
själva träningsprocessen.

e) Vad är grid search och hur hänger namnet “grid” och “search” ihop med
processen som genomförs? Läser vi dokumentationen: https://scikit-learn.
org/stable/modules/generated/sklearn.model_selection.GridSearchCV så
ser vi att hyperparametern refit har standardvärdet True, vad innebär
det?

- Grid search är en metod för att hitta de bästa hyperparametrarna för en modell.
- Grid: Värdena för de olika parametrarna, tex tree depth, som ska användas.
- Search: Letar efter den kombination av hyperparametrar som presterar bäst.

f) Vad är kategorisk data och hur hanteras det? I ditt svar, använd begreppen
nominal data, ordinal data, one-hot-encoding, dummy-variable-encoding
och ordinal encoding.

- Kategorisk data är data som delas in i kategorier, eller fasta grupper, och som inte mäts i siffror. Där har vi tex nominal data, som är kategorier som inte har någon inbördes ordning (hårfärg), samt ordinal data, som har en naturlig ordning (betyg).
För att kunna använda kategorisk data inom ML, måste vi omvandla datan till siffror. Detta eftersom att datorer bara förstår siffror. Då kan vi använda one-hot-encoding, dummy-variable-encoding eller ordinal encoding.

g) Vad är feature engineering för något?

- Det är ett sätt att ta fram meningsfull data som hjälper modellen att lära sig mönster effektivare. Detta görs genom domänkunskap och olika statistiska metoder för att exkludera data som inte är relevant.

h) Vad menas med principle of parsimony?

- Att den enklaste förklaringen oftast är den bästa förklaringen.

### 4.
Vad menas med att “en modell är en förenkling av verkligheten”?

- Att verkligheten är för komplex för en modell att hantera, så istället använder vi begränsad data och gör generaliseringar kring det.

### 5.
Vad menas med att en modell är “överanpassad” eller overfitted på engelska?

- Det innebär att modellen är för komplex för datan som används. Modellen tar hänsyn till för mycket "fel" data vilket leder till att den inte klarar generalisera kring datan.

### 6.
Högre är bättre i scikit-learn scoring, vad innebär det?

- Pga av hur scikit-learn har byggt sitt scoring-system, så betyder det att score alltid blir ett negativt värde. Eftersom vi vill ha ett score så nära 0 som möjligt, blir det att högre är bättre.
-2 är högre (närmare 0) än -10. 

### 7.
Vad är tvärsnittsdata, tidsseriedata och paneldata? Exemplifiera när respektive datakategori kan uppstå.

- Tvärsvittsdata: Data som samlats in under en specifik tidspunkt. Kan tex vara kundnöjdhet under en specifik vecka.
- Tidsseriedata: Data för en och samma mätpunkt vid flera tillfällen över tid. Kan tex vara temperaturen i ett specifikt rum, 3 ggr om dagen i 2 veckor.
- Paneldata: En kombination av ovan, där man följer flera samma mätpunkter över tid. Kan tex vara att följa temperaturen i 5 städer över en tid av 5 år.

### 9.
Generellt sett gäller det att högre är bättre i scikit-learn scoring, därför
används exempelvis scoring='neg_mean_squared_error'. Förklara logiken bakom detta, det vill säga att vi använder “negative” mean squared
error.

- Det är samma sak som RMSE, men att vi lägger ett - före och gör det negativt. Annledning till det är för att score-systemetet ska fungera likadant för de olika utvärderingsmåtten med scikit-learn, eftersom att här gäller högre är bättre. Kollar vi på tex r2 så gäller högre är bättre.
Iom att vi lägger ett negativt värde på RMSE så gäller samma där. Tex så är -2 större än -10.

### 11.
I koden nedan använder vi först test_size=0.2 och sedan test_size=0.25,
förklara varför det ger oss proportionerna 60-20-20.

- I första delen har vi test_size=0.2. Det betyder att vi delar datan 80% träningsdata och 20% testdata. Sen splittar vi datan. Då har vi test_size=0.25. Det betyder att vi tar de 80% av träningsdatan vi hade i steg 1, och tar 25% av det till valideringsdata. Eftersom att vi bara har 80% av datan att dela upp, så behöver vi ta 25% till validerningsdata. Detta för att behålla ungefär samma antal mätpunkter för både test och validering.
