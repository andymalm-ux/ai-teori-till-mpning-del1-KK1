## -- Faktafrågor --

### 1.
Vad är klustring för något? Ge några exempel på tillämpningsområden.
- Klustring är en oövervakad inlärningsmetod där man grupperar data i kluster (grupper) baserat på hur lika punkterna är varandra, utan att man i förväg har facit för vilken grupp varje punkt tillhör.

Idén är att punkter inom samma kluster ska vara så lika varandra som möjligt, medan punkter i olika kluster ska vara så olika som möjligt. Algoritmen hittar mönstren själv, till skillnad från klassificering där man tränar på data med kända etiketter.

- Kundsegmentering — gruppera kunder efter köpbeteende för riktad marknadsföring.
- Bildkomprimering/färgkvantisering — gruppera liknande pixelfärger.

### 2.
Förklara översiktligt hur K-means fungerar. Använd figur 6.3 (sidan 238)
och figur 6.4 (sidan 239) i din förklaring.
- Man bestämmer hur många kluster (k) man vill ha. Algoritmen placerar sedan k st "mittpunkter" och tilldelar varje datapunkt till den närmaste mittpunkten. Mittpunkterna flyttas sedan till mitten av sina tilldelade punkter, och man upprepar det här tills inget förändras längre. Resultatet blir grupper av punkter som ligger nära varandra.

Figur 6.3 visar fyra exempel på hur "sann" data kan se ut , alltså hur klustren egentligen är uppbyggda, innan man ens kört någon algoritm: runda, jämnstora moln (tre kluster), avlånga/diagonala kluster, kluster med olika spridning, och kluster med olika storlek.

Figur 6.4 visar vad som händer när K-means faktiskt körs på samma data, och det går fel i alla fyra fall:

Fel antal kluster — man bad om fel k, så en grupp klipps av och blandas ihop med en annan.
Icke-runda kluster — K-means antar att kluster är runda "moln". Är datan avlång/diagonal delar den upp fel, rakt genom det som egentligen är samma grupp.
Olika spridning — ett tätt, litet kluster "äter sig in i" ett stort utspritt kluster eftersom K-means bara tittar på avstånd till mittpunkten, inte hur tätt punkterna ligger.
Olika storlek — ett stort kluster stjäl punkter från ett litet grannkluster.

### 3.
Hur kan man välja vilket antal kluster som ska användas för en K-means
modell? Använd inertia, silhouette score och silhouette diagram i ditt svar.
- Det finns ingen given "rätt" k i K-means, så man använder några verktyg för att avgöra ett bra val: 
- Inertia (armbågsmetoden): Mäter hur "tighta" klustren är — hur nära punkterna ligger sitt eget centrum. Fler kluster ger alltid tightare grupper, så kurvan sjunker hela tiden. Man letar efter stället där kurvan slutar sjunka snabbt och istället planar ut — som en armbåge. Där är ett bra antal kluster.

- Silhouette score: Ett betyg (mellan -1 och 1) på hur bra varje punkt passar i sitt kluster jämfört med om den hade hamnat i ett annat. Man testar olika antal kluster och väljer det som ger högst genomsnittligt betyg.

- Silhouette-diagram: En bild som visar detta betyg för varje punkt, kluster för kluster. Man vill se ungefär lika breda staplar för alla kluster och att de flesta punkter har höga betyg — det tyder på bra, balanserade kluster. Ser man tunna staplar eller punkter med dåliga betyg, är antalet kluster troligen fel.

### 4.
Om du kollar på figur 6.10 på sidan 247, hur många kluster hade du valt
och varför? Är det en “exakt vetenskap” att välja antalet kluster?
-Jag hade valt 4 kluster.

I grafen ser man en brant nedgång från 1→2→3→4 kluster (inertian rasar från ~3500 till ~250), men efter 4 kluster planar kurvan ut. Fler kluster ger bara marginellt lägre inertia. Det är precis där armbågen (markerad i figuren) sitter: den punkt där man får "mest nytta per extra kluster" innan förbättringen avtar kraftigt.
Är det en exakt vetenskap? Nej, absolut inte. Armbågen är ofta subjektiv att avläsa. Det är en kombination av kvantitativa verktyg och sunt förnuft/domänkunskap, inte en formel som ger ett garanterat rätt svar.

### 5.
Hur tolkar man figur 6.13 på sidan 251?
- Varje "kniv" i diagrammet är ett kluster, och bredden på kniven visar hur många punkter som ligger i det klustret. Längden på strecken inuti visar hur bra varje enskild punkt passar i sitt kluster (längre = bättre). Det handlar inte bara om att maximera ett enda tal, utan om att väga ihop flera signaler (silhouette-värde + balans mellan klustren) och göra en avvägning. Återigen ingen exakt vetenskap, utan omdöme.
