## -- Faktafrågor --

### 1.
Vad menas med curse of dimensionality.
- Det syftar till problemet som kan uppstå när  ett dataset har väldigt många features. När antalet dimensioner öker så blir datan glesare vilket bidrar till att det blir svårare för modellen att hitta tydliga mönster.

### 2.
Vad är dimensionsreducering och varför görs det?
- Det innebär att man minskar antalet features i ett dataset, samtidigt som man försöker behålla så mycket viktig information som möjligt. Låt säga att vi har 100 features, där vissa features även har en korrelation till varandra. Dimensionsreducering kan då omvandla de 100 features till 10 nya dimensioner och behålla mycket av den ursprungliga informationen.

### 3.
Förklara översiktligt hur PCA fungerar. Använd figur 5.4 på sidan 224 i din förklaring.
- PCA hittar den riktning i datan där punkterna sprider sig mest, och gör den till en ny axel.

I figur 5.4: punkterna ligger ungefär på en diagonal. **c₁** är den riktningen, dit datan "pekar" mest. **c₂** står vinkelrätt mot c₁ och fångar det lilla som blir kvar.

Högra panelen visar varför det är smart: projicerar man punkterna på c₁ (översta raden) sprider de sig brett, man behåller det mesta av informationen. Projicerar man dem på c₂ istället (nedersta raden) klumpar de ihop sig nästan till en punkt, nästan ingen information kvar.

Så istället för att beskriva varje punkt med både x₁ och x₂, kan man ofta nöja sig med bara värdet längs c₁, man går från 2 tal till 1 tal per punkt, utan att förlora särskilt mycket.

### 5.
Stina påstår att man i maskininlärning alltid vill ha modeller som genomför
så bra prediktioner som möjligt. Kalle påstår att det inte riktigt stämmer
eftersom tid också är en viktig aspekt. Både för själva modellträningen
och för själva prediktionerna. Vad säger du?
- Kalle har rätt, och det finns fler aspekter än bara tid också. I praktiken finns det nästan alltid en avvägning (trade-off) mellan hur bra en modell presterar och andra faktorer.
- Träningstid: En mer komplex modell (t.ex. ett djupt neuralt nätverk eller en stor ensemble) kan ge marginellt bättre prediktioner, men kan ta timmar eller dagar att träna jämfört med sekunder för en enklare modell. Om man behöver iterera snabbt, testa idéer eller retränа ofta (t.ex. på ny data varje dag) blir det opraktiskt.
- Prediktionstid (inferenstid): Om modellen ska användas i realtid, t.ex. för att godkänna en betalning, styra en bil, eller ge sökresultat, kan en modell som tar 2 sekunder per prediktion vara oanvändbar även om den är 1 % mer träffsäker än en modell som svarar på 10 millisekunder.

### 6.
Efter att vi genomfört en PCA, vad händer med tolkningen av variablerna?
- Tolkningen av variablerna försämras kraftigt, det är faktiskt en av de största nackdelarna med PCA. Du byter tolkningsbarhet mot dimensionsreduktion. Det är en avvägning, precis som Kalles poäng om tid/prestanda i föregående fråga: PCA ger dig ofta bättre prestanda/effektivitet, men till priset av att variablerna inte längre går att förstå på samma direkta sätt som innan.
