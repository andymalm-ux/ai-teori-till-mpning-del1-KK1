### 1. 
Vad kännetecknar regressionsproblem? Ge några exempel på tillämpningsområden.
- Regressionsproblem kännetecknas av att man har en beroende variabel, y, och en eller flera oberoende variabler, x. Med hjälp av x så försöker vi prediktera y.
Det kan tex vara att prediktera en persons lön (y) baserat på personens ålder (x), eller prediktera huspris (y) baserat på storlek, ålder, plats (x). 

### 2. 
Förklara utvärderingsmåtten RMSE, MSE och MAE.
- MAE tar det absoluta felet för varje observation och räknar sedan ut ett medelvärde.
- MSE tar felen och kvadrerar dom för att sedan räkna ut medelvärdet.
- RMSE är roten ur MSE.


### 3. 
Om vi ska rangordna olika modeller, spelar det någon roll om RMSE eller MSE används? Varför?
- Nej det spelar ingen roll om vi använder RMSE eller MSE, då båda ger oss samma rangordning. Skillnaden mellan de 2 är att RMSE tar roten ur det värde som MSE får. MSE är på så vis däremot lite snabbare,
då det inte behöver göra uträkning för roten ur.

### 4. 
Förklara mycket översiktligt vad gradient descent är.
- Det är en optimeringsalgoritm som finns i biblioteket scikit-learn. Algoritmen gör att ML-modellerna försöker lära sig de parametrar som gör att skillnaderna mellan de sanna värdena, och de predikerade värdena,
blir så liten som möjligt.

### 5. 
Vad är the bias variance trade-off? Varför är mer komplexa modeller inte alltid bättre?
- Det är balansen mellan en modells enkelhet, bias, och dess känslighet, variance, gentemot träningsdata.
Om du har en för komplex modell baserat på datan, finns risken för att modellen gör en sk overfitting. Det innebär att modellen fångar upp för mycket små avvikelser i datan,
istället för att generalisera.

### 6. 
Förklara översiktligt hur respektive modell fungerar:

a) Linjär regression
- Med linjär regression försöker vi hitta ett linjärt samband mellan indata X, och ett målvärde y.

b) Ridge regression
- Ridge regression är samma som linjär regression, med tillägget regularisering. Det innebär att modellen lägger till ett straff för stora koefficienter. Detta tenderar att göra modellen mindre känslig för brus och överanpassning.

c) Lasso regression
- Lasso regression är lik ridge regression, men straffar på ett annat sätt. En egenskap är också att modellen kan välja bort mindre relevanta variabler automatiskt.

d) Elastic net
- Elastic net är en kombination av ridge och lasso. Lämplig att använda när man tex har många features och om det finns en stark korrelation mellan features.

e) Support vector machines
- SVM används för att separera olika grupper av data på ett bra sätt. Det kan tex vara att man vill kategorisera bilar efter billig eller dyr. Modellen försöker hitta en gräns där avståndet till de närmsta datapunkterna är så stort som möjligt.

f) Beslutsträd
- Fungerar ungefär som en serie if/else-frågor. 

g) Ensemble learning
- Är egentligen inte en modell, utan ett sätt att bygga modeller. Man tränar och kombinerar resultatet från flera modeller.

h) Random forest
- Är en ensemble av många beslutsträd. Istället för att tränas på ett träd så tränas modellen på flera träd, där varje träd tränas på slumpmässig del av data och/eller features. Sedan kombineras alla träden till en modell.

### 7.
Vad menas med white box modeller och black box modeller?
- Whitebox modeller kallas de modeller som generellt är intuitiva förklara och där det är lätt  att se varför en modell gör en viss prediktion. Exempel, Beslutsträd.
- Blackbox modeller kallas de modeller där man generellt kan följa beräkningarna och prediktionerna matematiskt, men dom är svåra att förklar i ord. Exempel, Random forest.

### 8.
Vad är skillnaden mellan bagging och pasting?
- Skillnaden mellan de 2 är att när man skapar ett nytt slumpmässigt dataset (från ett givet dataset) med bagging, så kan samma observationspunkter förekomma flera gånger. Det gör dom inte när man skapar datasetet med
pasting. I övrigt gör dom samma sak. Dom skapar ett nytt slumpmässigt dataset utifrån ett givet dataset.

### 10.
Förklara hur man kan tolka figur 3.13 på sidan 140. Hur hänger den ihop med figur 3.14 på sidan 141?
- Figur 3.13 (trädet): Ett beslutsträd fungerar som ett frågeschema. Man börjar högst upp och ställer en fråga om datan (t.ex. "är X2 mindre än 0.438?"). Beroende på svaret (True/False) går man vidare till nästa fråga, tills man når en "låda" längst ner (ett löv) som ger en förutsägelse (value), ett genomsnittligt värde baserat på de träningspunkter (samples) som hamnade där.

I det här trädet ställs först frågan om X2, och beroende på svar ställs en till fråga (antingen om X2 igen, eller om X1), så till slut delas datan upp i fyra grupper, var och en med sitt eget förutsagda värde.

Kopplingen till figur 3.14: Den bilden är en visuell karta över exakt samma fyra grupper som trädet skapade, ritade på ett diagram med X1 och X2 som axlar. Varje färgat rektangulärt område motsvarar ett löv i trädet, samma gränsvärden (X2 = 0.438, X2 = -0.755, X1 = -0.654) syns här som räta linjer som delar upp planet i fyra rutor. Färgen visar det förutsagda värdet (blått = lågt, rött = högt), och man ser att det matchar värdena i trädets löv perfekt.

### 11.
Förklara vad 𝑅2 är för mått.
- Det är ett mått som, till skillnad från RMSE och MAE, används för att mäta hur bra en modell kan förklara variation i data. Säg att vi ska förutsäga huspriser. Huspriser skiljer sig åt beroende på olika variablar, som läge, storlek, ålder osv. Med hjälp av dom variablarna så kan modellen förklara variationen i pris. R2 mäter hur bra den gör det utifrån de variablar den har fått.
